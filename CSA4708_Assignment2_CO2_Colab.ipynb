## Task 1 – Deep Feed-Forward Network Architecture (5 Marks)

A crop-yield prediction system receives four numerical inputs: rainfall, temperature, soil moisture and fertilizer quantity. The network contains two hidden layers with 8 and 4 neurons.

### (a) Network design
- **Input layer:** 4 neurons (one for each input feature).
- **Hidden layer 1:** 8 neurons.
- **Hidden layer 2:** 4 neurons.
- **Output layer:** 1 neuron (predicted crop yield).

Architecture: **4 → 8 → 4 → 1**

### (b) Weight and bias dimensions
Using column-vector notation, each layer computes \(z^{(l)}=W^{(l)}a^{(l-1)}+b^{(l)}\).

| Layer connection | Weight matrix | Bias vector |
|---|---:|---:|
| Input → Hidden 1 | 8 × 4 | 8 × 1 |
| Hidden 1 → Hidden 2 | 4 × 8 | 4 × 1 |
| Hidden 2 → Output | 1 × 4 | 1 × 1 |

### (c) Forward propagation
The four input values are passed to the first hidden layer. Each layer multiplies its input by weights, adds biases and applies an activation function. The final output neuron produces the predicted crop yield.

For example:
\[
a^{(1)}=ReLU(W^{(1)}x+b^{(1)})
\]
\[
a^{(2)}=ReLU(W^{(2)}a^{(1)}+b^{(2)})
\]
\[
\hat y=W^{(3)}a^{(2)}+b^{(3)}
\]

A linear output is suitable for a continuous regression target such as crop yield.

```python
import numpy as np

# Demonstrate the dimensions and one forward pass
rng = np.random.default_rng(42)
x = np.array([[0.6], [0.7], [0.4], [0.5]])  # normalized input features
W1, b1 = rng.normal(size=(8, 4)), rng.normal(size=(8, 1))
W2, b2 = rng.normal(size=(4, 8)), rng.normal(size=(4, 1))
W3, b3 = rng.normal(size=(1, 4)), rng.normal(size=(1, 1))

relu = lambda z: np.maximum(0, z)
a1 = relu(W1 @ x + b1)
a2 = relu(W2 @ a1 + b2)
y_hat = W3 @ a2 + b3

print("W1, b1:", W1.shape, b1.shape)
print("W2, b2:", W2.shape, b2.shape)
print("W3, b3:", W3.shape, b3.shape)
print("Predicted crop yield (illustrative):", float(y_hat[0, 0]))
```

## Task 2 – Forward Propagation and Activation Functions (2 Marks)

Given inputs 0.6, 0.8 and 0.5; weights 0.4, 0.3 and 0.2; and bias 0.1.

### (a) Weighted sum
\[
z=(0.6\times0.4)+(0.8\times0.3)+(0.5\times0.2)+0.1
\]
\[
z=0.24+0.24+0.10+0.10=0.68
\]

### (b) ReLU activation
\[
ReLU(z)=\max(0,z)=\max(0,0.68)=0.68
\]

**Interpretation:** Since the weighted sum is positive, ReLU returns 0.68 unchanged. ReLU converts negative values to zero and keeps positive values.

```python
inputs = np.array([0.6, 0.8, 0.5])
weights = np.array([0.4, 0.3, 0.2])
bias = 0.1
weighted_sum = np.dot(inputs, weights) + bias
relu_output = max(0, weighted_sum)
print("Weighted sum:", weighted_sum)
print("ReLU output:", relu_output)
```

## Task 3 – Gradient-Based Learning and Loss Calculation (6 Marks)

The predicted crop yield is 42 units and the actual yield is 50 units.

### (a) Suitable loss function
A loss function measures the difference between predicted and actual values. For regression, Mean Squared Error (MSE) is commonly suitable because it squares prediction errors and penalizes larger errors more strongly.

### (b) Squared error
\[
L=(y-\hat y)^2=(50-42)^2=8^2=64
\]

### (c) Interpretation
The squared error is **64 squared units**. The prediction is 8 units below the actual yield. A smaller loss indicates a prediction closer to the actual value.

### (d) Gradient-based learning
1. Perform forward propagation to obtain a prediction.
2. Calculate the loss.
3. Differentiate the loss with respect to each trainable weight and bias.
4. Use the gradients to adjust parameters in a direction that reduces loss.
5. Repeat over batches and epochs.

```python
actual_yield = 50
predicted_yield = 42
squared_error = (actual_yield - predicted_yield) ** 2
print("Squared error loss:", squared_error)
```

## Task 4 – Gradient Descent and Weight Update (4 Marks)

Given weight \(w=0.5\), gradient \(\frac{dL}{dw}=0.8\), and learning rate \(\eta=0.1\).

### (a) Updated weight
Gradient descent rule:
\[
w_{new}=w_{old}-\eta\frac{dL}{dw}
\]
\[
w_{new}=0.5-(0.1\times0.8)=0.5-0.08=\boxed{0.42}
\]

### (b) Why the opposite direction?
The gradient points in the direction of the steepest increase in loss. Subtracting it moves the weight in the direction of steepest local decrease.

### (c) Learning-rate effects
- **Extremely small:** updates are tiny, so training becomes very slow.
- **Extremely large:** updates may overshoot the minimum, fluctuate or diverge.
- **Suitable rate:** supports stable and reasonably fast convergence.

```python
weight = 0.5
gradient = 0.8
learning_rate = 0.1
updated_weight = weight - learning_rate * gradient
print("Updated weight:", updated_weight)
```

## Task 5 – Back-Propagation Through Multiple Layers (3 Marks)

### (a) How backpropagation works
Backpropagation calculates how much each weight and bias contributes to the prediction error. It starts at the output layer, computes the output error and propagates error information backward through the hidden layers.

### (b) Role of the chain rule
The chain rule calculates the derivative of the loss with respect to an earlier layer's parameters by multiplying derivatives along the computational path. For a weight in an early layer, the gradient includes the effect of that weight on later activations and finally on the loss.

### (c) Weight updates
After gradients are calculated, an optimizer such as gradient descent updates each parameter:
\[
W_{new}=W_{old}-\eta\frac{\partial L}{\partial W}
\]
This process is repeated to reduce prediction error.

## Task 6 – Activation Functions and Vanishing/Exploding Gradients (3 Marks)

### (a) Vanishing gradients
Sigmoid and tanh can become saturated for very large positive or negative inputs. Their derivatives then become very small. During backpropagation, multiplying many small derivatives can make gradients in early layers almost zero. Those layers learn very slowly.

Exploding gradients occur when repeated derivative multiplications produce very large values, causing unstable updates.

### (b) Comparison

| Function | Output range | Advantages | Limitations |
|---|---|---|---|
| Sigmoid | 0 to 1 | Smooth; useful for binary output probabilities | Saturates; vanishing gradients; not zero-centred |
| Tanh | -1 to 1 | Zero-centred output | Can also saturate and cause vanishing gradients |
| ReLU | 0 to infinity | Simple; helps reduce saturation for positive inputs | Neurons can become inactive for negative inputs |

### (c) Effect on training
ReLU is commonly used in hidden layers because its derivative is 1 for positive inputs, helping gradient flow. The output activation should match the task: linear for regression and sigmoid/softmax for classification.

```python
import matplotlib.pyplot as plt

z = np.linspace(-6, 6, 300)
sigmoid = 1 / (1 + np.exp(-z))
tanh = np.tanh(z)
relu_values = np.maximum(0, z)

plt.figure(figsize=(8, 4))
plt.plot(z, sigmoid, label="Sigmoid")
plt.plot(z, tanh, label="Tanh")
plt.plot(z, relu_values, label="ReLU")
plt.title("Activation Function Comparison")
plt.xlabel("Input")
plt.ylabel("Output")
plt.grid(True)
plt.legend()
plt.show()
```

## Task 7 – Back-Propagation and Optimization Strategy (4 Marks)

### (a) Learning rate
The learning rate controls the size of each parameter update. A small rate can make loss decrease slowly. A large rate can cause loss to fluctuate or fail to converge.

### (b) One training iteration
1. Forward propagation produces predictions.
2. The loss function measures prediction error.
3. Backpropagation computes gradients using the chain rule.
4. The optimizer updates weights and biases.
5. The next iteration uses the updated parameters.

### (c) Improving convergence
- **Slowly decreasing loss:** cautiously increase the learning rate or use an adaptive optimizer such as Adam; check feature scaling and model capacity.
- **Fast but fluctuating loss:** reduce the learning rate, use mini-batches, consider momentum/Adam and monitor validation loss.

### (d) Loss-curve monitoring
The training and validation loss curves show whether the model is learning, converging, overfitting or unstable. A widening gap between training and validation loss can indicate overfitting.

```python
# Illustrative loss curves (not measured training results)
epochs = np.arange(1, 21)
slow_loss = 1.8 * np.exp(-0.035 * epochs) + 0.15
fluctuating_loss = 1.8 * np.exp(-0.12 * epochs) + 0.12 + 0.08 * np.sin(epochs * 1.8)

plt.figure(figsize=(8, 4))
plt.plot(epochs, slow_loss, marker="o", label="Slow decrease")
plt.plot(epochs, fluctuating_loss, marker=".", label="Fast but fluctuating")
plt.xlabel("Epoch")
plt.ylabel("Illustrative training loss")
plt.title("Examples of Optimization Behaviour")
plt.grid(True)
plt.legend()
plt.show()
```

## Task 8 – Regularization and Overfitting Control (4 Marks)

The model has 99% training accuracy but only 78% validation accuracy.

### (a) Learning problem
This is likely **overfitting**. The model performs very well on training data but substantially worse on unseen validation data, suggesting that it has learned training-specific patterns.

### (b) L1 and L2 regularization
- **L1 regularization:** adds the sum of absolute weight values to the loss. It can encourage some weights to become exactly zero, producing a sparse model.
- **L2 regularization:** adds the sum of squared weights to the loss. It discourages excessively large weights and helps control model complexity.

### (c) L2 regularized loss
\[
L_{total}=L_{data}+\lambda\sum_i w_i^2
\]
Here, \(L_{data}\) is the original task loss and \(\lambda\) controls the strength of the penalty. A larger \(\lambda\) applies stronger weight shrinkage; an excessively large value can cause underfitting.

### (d) Dropout
Dropout randomly disables a fraction of neurons during training. This reduces reliance on particular neurons and encourages more robust feature representations, which may improve generalization.

```python
# Numerical illustration of L2 penalty
weights_example = np.array([0.5, -0.2, 0.8])
data_loss = 0.40
lambda_l2 = 0.01
l2_penalty = lambda_l2 * np.sum(weights_example ** 2)
regularized_loss = data_loss + l2_penalty
print("Data loss:", data_loss)
print("L2 penalty:", l2_penalty)
print("Regularized loss:", regularized_loss)
```

## Task 9 – Convolution Operation and Feature Extraction (3 Marks)

### (a) Why convolution is suitable for images
Images contain local spatial patterns such as edges, spots and textures. Convolutional layers use small filters that scan local regions and detect these patterns. The same filter weights are reused across the image, reducing parameters compared with a fully connected layer on raw pixels.

### (b) Numerical convolution example
Consider the image patch and filter:

\[
I=\begin{bmatrix}1&2&0\\0&1&3\\2&1&0\end{bmatrix},
\quad
K=\begin{bmatrix}1&0\\0&-1\end{bmatrix}
\]

At the top-left position:
\[
(1\times1)+(2\times0)+(0\times0)+(1\times-1)=0
\]

Repeating at each valid position gives the feature map shown by the code below. This filter responds to local intensity differences; the values indicate the filter response, not a disease diagnosis.

### (c) Filter, stride and padding
- **Filter/kernel:** small matrix of learnable values used to detect patterns.
- **Stride:** number of pixels the filter moves at each step. Larger stride reduces output dimensions.
- **Padding:** adds values (often zeros) around the image boundary. It can preserve spatial dimensions and retain edge information.

```python
image_patch = np.array([[1, 2, 0],
                        [0, 1, 3],
                        [2, 1, 0]])
kernel = np.array([[1, 0],
                   [0, -1]])

feature_map = np.zeros((2, 2), dtype=int)
for i in range(2):
    for j in range(2):
        feature_map[i, j] = np.sum(image_patch[i:i+2, j:j+2] * kernel)

print("Input image patch:\n", image_patch)
print("Filter:\n", kernel)
print("Convolution feature map:\n", feature_map)
```

## Task 10 – CNN Architecture and Pooling (4 Marks)

Given architecture:

**Input Image → Convolution → ReLU → Pooling → Convolution → ReLU → Pooling → Fully Connected Layer → Output**

### (a) Component functions
1. **Input image:** supplies pixel values.
2. **Convolution:** applies filters to extract local features.
3. **ReLU:** introduces non-linearity by replacing negative values with zero.
4. **Pooling:** reduces spatial dimensions while retaining prominent information.
5. **Second convolution and ReLU:** learn more complex patterns from earlier feature maps.
6. **Second pooling:** further reduces feature-map dimensions.
7. **Fully connected layer:** combines extracted features for classification.
8. **Output:** produces class scores or probabilities, such as healthy and diseased.

### (b) Pooling example
A 2 × 2 max-pooling operation takes the maximum value from each 2 × 2 region. For example, the region [[1, 3], [2, 4]] becomes 4. This reduces width and height.

### (c) Feature-map effects
- Larger filters examine a wider local region.
- Larger stride reduces output width and height.
- More filters produce more output channels and can detect more feature types, but increase computation and parameters.

### (d) Hierarchical feature extraction
Early convolutional layers may learn edges and simple textures. Deeper layers combine these into shapes, spots and more complex leaf patterns.

```python
feature_map_demo = np.array([[1, 3, 2, 4],
                             [5, 6, 1, 2],
                             [0, 2, 7, 3],
                             [4, 1, 8, 9]])

# 2x2 max pooling with stride 2
pooled = np.array([
    [feature_map_demo[i:i+2, j:j+2].max()
     for j in range(0, 4, 2)]
    for i in range(0, 4, 2)
])
print("Feature map:\n", feature_map_demo)
print("After 2x2 max pooling:\n", pooled)
```

## Task 11 – Comparative Evaluation of Feed-Forward and CNN Models (4 Marks)

### (a) Feed-forward network for structured numerical data
Rainfall, temperature, soil moisture and fertilizer quantity are already represented as numerical features in rows and columns. A feed-forward network can directly process these fixed-length feature vectors and predict a continuous crop-yield value.

### (b) CNN for images
A CNN preserves spatial relationships between neighbouring pixels and learns local visual patterns such as leaf edges, spots and textures. It is therefore suitable for leaf-image classification.

### (c) Comparison

| Aspect | Model A: Feed-forward network | Model B: CNN |
|---|---|---|
| Input representation | Numerical feature vector | Image tensor (height × width × channels) |
| Feature extraction | Dense layers learn feature combinations | Convolution filters learn spatial features |
| Parameter organization | Dense connections between adjacent layers | Shared filter weights plus dense layers |
| Computation | Usually lower for a small number of numerical features | Often higher due to image and feature-map processing |
| Main strength | Structured numerical prediction | Spatial image classification |

### (d) Recommendation

## Task 12 – Design Challenge: Optimize and Redesign the System (8 Marks)

The original system was trained on 10,000 crop records. The new requirement is 500,000 crop records and 100,000 leaf images. Training accuracy is high, validation performance is lower, and training takes a long time.

### A. Impact Analysis (2 Marks)

**(a) Effect of larger data volume and images**
- 500,000 numerical records require a scalable data pipeline, efficient batching, sufficient memory and careful training-time management.
- 100,000 images require image storage, preprocessing, augmentation and a CNN architecture capable of learning spatial patterns.
- Numerical and image data have different structures, so one dense network is not the most suitable single architecture for both tasks.

**(b) Limitations of applying a feed-forward network directly to images**
Flattening an image into one long vector removes its explicit 2D spatial structure. Dense layers may require many parameters and do not naturally exploit local patterns or shared filters. A CNN is more suitable for image feature extraction.

### B. Evaluation of Existing Design (2 Marks)

**(a) Existing model**
High training accuracy and lower validation performance indicate likely overfitting. Long training time also suggests that the current design or training pipeline may not scale efficiently to the larger requirement.

**(b) Is simply increasing depth and neurons best?**
No. A larger network may increase computation and overfitting. The architecture, regularization, data quality, preprocessing, optimizer, learning rate and validation performance should be considered together.

### C. Adaptation or Redesign (2 Marks)

**(a) Proposed architecture**
- **Numerical crop data:** input features → dense layer (64, ReLU) → dropout → dense layer (32, ReLU) → one linear output for crop-yield regression.
- **Leaf images:** image input → convolution + ReLU + pooling blocks → global average pooling or flattening → dense layer → softmax output for healthy/diseased classification.

**(b) Optimization and regularization**
- Start with Adam for adaptive parameter updates; compare against SGD with momentum.
- Use a learning-rate schedule, such as reducing the rate when validation loss stops improving.
- Apply dropout and L2 regularization to reduce overfitting.
- Use batch normalization in convolutional/dense blocks where appropriate.
- Use ReLU in hidden layers; use linear output for regression and softmax for multi-class classification (sigmoid for binary classification).

### Additional design considerations
**Optimizer comparison**
- **SGD:** simple updates using gradients; can be sensitive to learning rate.
- **Mini-batch SGD:** uses a subset of training examples per update; balances noisy updates and computation.
- **Momentum:** accumulates a moving direction from past gradients to smooth updates and accelerate progress.
- **Adam:** adapts learning rates per parameter using moving averages of gradients and squared gradients.
- **RMSProp:** adapts updates using a moving average of squared gradients.

**Suggested learning-rate schedule:** begin with a moderate learning rate, monitor validation loss, and reduce the rate when validation loss plateaus. Early stopping can stop training when validation performance no longer improves.

**Label-preserving image augmentation:** small rotations, horizontal flips when leaf orientation is not label-dependent, modest zoom, translation, and brightness/contrast changes. Avoid transformations that remove disease symptoms or change the label.

**CNN architecture comparison**
- **LeNet:** small and simple; useful for small, low-complexity image tasks.
- **AlexNet:** deeper and computationally heavier than LeNet; historically important for larger image classification.
- **ResNet:** residual connections help train deeper networks.
- **EfficientNet:** balances accuracy and computational cost through compound scaling.

For a limited compute budget, begin with a compact CNN or a small pretrained EfficientNet/ResNet variant, then compare validation performance and training time. The best choice depends on image resolution, dataset quality and available hardware.

## Practical Experiment – Train and Compare Models (Synthetic Demonstration)

This section creates a small synthetic numerical dataset and a synthetic image dataset. It demonstrates the training workflow and evaluation metrics. The data are randomly generated and **must not be treated as real crop or disease data**.

The code uses TensorFlow/Keras. If TensorFlow is unavailable in the current runtime, run the notebook in Google Colab with a Python 3 runtime.

```python
# Install only if needed in a fresh environment
import importlib.util
if importlib.util.find_spec("tensorflow") is None:
    %pip -q install tensorflow
```

```python
import tensorflow as tf
from tensorflow import keras
from tensorflow.keras import layers
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.metrics import mean_absolute_error, mean_squared_error, accuracy_score, precision_score, recall_score, f1_score
import time

SEED = 42
np.random.seed(SEED)
tf.random.set_seed(SEED)
print("TensorFlow version:", tf.__version__)
```

```python
# Synthetic numerical crop data: 4 features and a continuous yield target
rng = np.random.default_rng(SEED)
n_samples = 3000
X_num = rng.normal(size=(n_samples, 4)).astype("float32")
y_num = (30 + 5*X_num[:, 0] - 3*X_num[:, 1] + 4*X_num[:, 2] +
         2*X_num[:, 3] + rng.normal(0, 1.5, n_samples)).astype("float32")

X_train, X_test, y_train, y_test = train_test_split(
    X_num, y_num, test_size=0.2, random_state=SEED
)
scaler = StandardScaler()
X_train = scaler.fit_transform(X_train).astype("float32")
X_test = scaler.transform(X_test).astype("float32")

def build_ffn():
    model = keras.Sequential([
        layers.Input(shape=(4,)),
        layers.Dense(64, activation="relu"),
        layers.Dropout(0.25),
        layers.Dense(32, activation="relu"),
        layers.Dense(1)
    ])
    model.compile(optimizer=keras.optimizers.Adam(learning_rate=0.001),
                  loss="mse", metrics=["mae"])
    return model

ffn = build_ffn()
start = time.time()
history_num = ffn.fit(X_train, y_train, validation_split=0.2,
                      epochs=25, batch_size=32, verbose=0,
                      callbacks=[keras.callbacks.EarlyStopping(
                          monitor="val_loss", patience=4, restore_best_weights=True)])
ffn_time = time.time() - start
pred_num = ffn.predict(X_test, verbose=0).ravel()
print("Numerical model training time (seconds):", round(ffn_time, 2))
print("Test MAE:", round(mean_absolute_error(y_test, pred_num), 3))
print("Test RMSE:", round(np.sqrt(mean_squared_error(y_test, pred_num)), 3))
```

```python
plt.figure(figsize=(8, 4))
plt.plot(history_num.history["loss"], label="Training loss")
plt.plot(history_num.history["val_loss"], label="Validation loss")
plt.title("Feed-Forward Network: Loss")
plt.xlabel("Epoch")
plt.ylabel("MSE loss")
plt.grid(True)
plt.legend()
plt.show()
```

```python
# Synthetic leaf images: two classes with simple visual patterns.
# Class 0: mostly green; Class 1: green with brown spots.
n_images, height, width, channels = 1200, 32, 32, 3
X_img = np.zeros((n_images, height, width, channels), dtype="float32")
y_img = np.zeros(n_images, dtype="int32")

for i in range(n_images):
    img = np.zeros((height, width, 3), dtype="float32")
    img[..., 1] = 0.55 + rng.normal(0, 0.04, (height, width))  # green channel
    img[..., 0] = 0.12 + rng.normal(0, 0.02, (height, width))
    img[..., 2] = 0.08 + rng.normal(0, 0.02, (height, width))
    label = i % 2
    if label == 1:
        for _ in range(5):
            r, c = rng.integers(3, 29, size=2)
            img[max(0,r-2):r+2, max(0,c-2):c+2, 0] = 0.55
            img[max(0,r-2):r+2, max(0,c-2):c+2, 1] = 0.25
    X_img[i] = np.clip(img, 0, 1)
    y_img[i] = label

Xi_train, Xi_test, yi_train, yi_test = train_test_split(
    X_img, y_img, test_size=0.2, random_state=SEED, stratify=y_img
)
print("Training image shape:", Xi_train.shape)
print("Test image shape:", Xi_test.shape)
```

```python
# Display examples from the synthetic image dataset
fig, axes = plt.subplots(1, 4, figsize=(9, 3))
for ax, idx in zip(axes, [0, 1, 2, 3]):
    ax.imshow(Xi_train[idx])
    ax.set_title("Diseased" if yi_train[idx] == 1 else "Healthy")
    ax.axis("off")
plt.suptitle("Synthetic Leaf Images (Illustrative Only)")
plt.tight_layout()
plt.show()
```

```python
def build_cnn():
    model = keras.Sequential([
        layers.Input(shape=(32, 32, 3)),
        layers.Conv2D(16, 3, padding="same", use_bias=False),
        layers.BatchNormalization(),
        layers.ReLU(),
        layers.MaxPooling2D(),
        layers.Conv2D(32, 3, padding="same", use_bias=False),
        layers.BatchNormalization(),
        layers.ReLU(),
        layers.MaxPooling2D(),
        layers.Conv2D(64, 3, padding="same", activation="relu"),
        layers.GlobalAveragePooling2D(),
        layers.Dropout(0.3),
        layers.Dense(1, activation="sigmoid")
    ])
    model.compile(optimizer=keras.optimizers.Adam(learning_rate=0.001),
                  loss="binary_crossentropy", metrics=["accuracy"])
    return model

cnn = build_cnn()
start = time.time()
history_img = cnn.fit(Xi_train, yi_train, validation_split=0.2,
                      epochs=12, batch_size=32, verbose=0,
                      callbacks=[keras.callbacks.EarlyStopping(
                          monitor="val_loss", patience=3, restore_best_weights=True)])
cnn_time = time.time() - start
prob_img = cnn.predict(Xi_test, verbose=0).ravel()
pred_img = (prob_img >= 0.5).astype(int)
print("CNN training time (seconds):", round(cnn_time, 2))
print("Test accuracy:", round(accuracy_score(yi_test, pred_img), 3))
print("Precision:", round(precision_score(yi_test, pred_img, zero_division=0), 3))
print("Recall:", round(recall_score(yi_test, pred_img, zero_division=0), 3))
print("F1-score:", round(f1_score(yi_test, pred_img, zero_division=0), 3))
```

```python
plt.figure(figsize=(8, 4))
plt.plot(history_img.history["loss"], label="Training loss")
plt.plot(history_img.history["val_loss"], label="Validation loss")
plt.title("CNN: Loss")
plt.xlabel("Epoch")
plt.ylabel("Binary cross-entropy")
plt.grid(True)
plt.legend()
plt.show()
```
