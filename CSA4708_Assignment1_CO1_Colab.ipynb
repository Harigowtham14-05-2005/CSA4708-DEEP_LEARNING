## Task 1 – Patient Data Representation and Normalization (5 Marks)
Raw features have different units and ranges. Large-scale features can dominate distances and affect gradient-based optimization. Min–max normalization is x'=(x−min)/(max−min). The assignment does not specify the three patients' values, so the following values are illustrative. Normalize using training-set statistics only in a real model.

```python
features = ["Age","Blood pressure","Glucose","Cholesterol"]
X = np.array([[25,110,90,170],[45,130,120,210],[65,150,160,250]],float)
Xn = (X-X.min(axis=0))/(X.max(axis=0)-X.min(axis=0))
print("Original matrix:\n",pd.DataFrame(X,columns=features))
print("Normalized matrix:\n",pd.DataFrame(Xn,columns=features))
print("Column minima:",Xn.min(axis=0),"Column maxima:",Xn.max(axis=0))
fig,ax=plt.subplots(1,2,figsize=(11,4))
ax[0].plot(features,X.T,marker="o"); ax[0].set_title("Raw features"); ax[0].tick_params(axis="x",rotation=30); ax[0].grid()
ax[1].plot(features,Xn.T,marker="o"); ax[1].set_title("Min-max normalized"); ax[1].tick_params(axis="x",rotation=30); ax[1].grid()
plt.tight_layout(); plt.show()
```

## Task 3 – Probability and Conditional Probability (6 Marks)
Let D denote disease and G denote elevated glucose. Given 160 diseased patients with elevated glucose and 120 non-diseased patients with elevated glucose, among 1,000 patients:

- P(G)=(160+120)/1000=0.28
- P(D|G)=160/280=0.5714
- P(G|D)=160/200=0.80

Conditional probability can help prioritize further assessment, but elevated glucose alone is not a diagnosis.

```python
total=1000; diseased=200
disease_high=160; no_disease_high=120
high=disease_high+no_disease_high
print("P(G) =",high/total)
print("P(D | G) =",disease_high/high)
print("P(G | D) =",disease_high/diseased)
display(pd.DataFrame([[160,40],[120,680]],index=["Disease","No disease"],columns=["Elevated glucose","Normal glucose"]))
```

## Task 4 – Bayesian Statistics for Disease Classification (4 Marks)
Given P(D)=0.10, P(+|D)=0.90, P(+|not D)=0.05 and P(not D)=0.90:

P(D|+) = [P(+|D)P(D)] / [P(+|D)P(D)+P(+|not D)P(not D)] = 0.09/0.135 = 0.6667.

The prior is the probability before the test; likelihood describes test evidence under each status; posterior is the updated probability after the positive result. Bayesian reasoning makes uncertainty explicit.

```python
prior=.10; sensitivity=.90; false_positive=.05
numerator=sensitivity*prior
denominator=numerator+false_positive*(1-prior)
print("P(+) =",denominator)
print("P(D | +) =",numerator/denominator)
```

## Task 5 – Entropy and Information Gain (3 Marks)
For 40 high-risk and 60 low-risk records, H(Y)=−Σ p log₂p = −[0.4log₂0.4+0.6log₂0.6] ≈ 0.971 bits. Entropy measures class uncertainty. Information gain is parent entropy minus weighted child entropy after splitting by a feature such as glucose.

```python
p=np.array([.4,.6])
H=-np.sum(p*np.log2(p))
print("Entropy:",H,"bits")
def entropy_counts(a,b):
    q=np.array([a,b],float)/(a+b)
    return -np.sum(q[q>0]*np.log2(q[q>0]))
# Illustrative split only; not supplied by the assignment
child=(50/100)*entropy_counts(30,20)+(50/100)*entropy_counts(10,40)
print("Illustrative weighted child entropy:",child)
print("Illustrative information gain:",H-child)
```

## Task 10 – Unsupervised Learning and Patient Grouping (4 Marks)
Unlabelled patient records are suitable for unsupervised learning. K-means chooses K centroids, assigns each patient to the nearest centroid, recalculates centroids as cluster means, and repeats until assignments stabilize. Distance measures similarity in normalized feature space. Clusters can guide further clinical investigation, but do not replace diagnosis.

```python
from sklearn.cluster import KMeans
from sklearn.preprocessing import StandardScaler
rng=np.random.default_rng(10)
g1=rng.normal([30,115,95,180],[5,8,10,15],(30,4))
g2=rng.normal([55,135,135,220],[6,8,12,18],(30,4))
g3=rng.normal([70,150,165,250],[5,7,12,15],(30,4))
demo=np.vstack([g1,g2,g3]) # illustrative synthetic data
scaler=StandardScaler(); Z=scaler.fit_transform(demo)
km=KMeans(n_clusters=3,random_state=42,n_init=10)
labels=km.fit_predict(Z)
print("Cluster sizes:",np.bincount(labels))
print("Cluster centers (original units):\n",scaler.inverse_transform(km.cluster_centers_))
plt.scatter(Z[:,0],Z[:,2],c=labels); plt.xlabel("Standardized age"); plt.ylabel("Standardized glucose")
plt.title("Illustrative K-means patient groups"); plt.grid(); plt.show()
```
